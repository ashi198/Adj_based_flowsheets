```python
import pickle
import numpy as np


def dump_pickle_flowsheets_txt(pickle_path, txt_path):
    """
    Read a pickle file containing a list of flowsheet dictionaries
    and dump all flowsheets into a human-readable txt file.
    """

    with open(pickle_path, "rb") as f:
        flowsheets = pickle.load(f)

    with open(txt_path, "w") as f:

        f.write(f"Total flowsheets: {len(flowsheets)}\n")
        f.write("=" * 120 + "\n\n")

        for idx, x in enumerate(flowsheets):

            pi = x["problem_instance"]
            graph = x["graph"]
            identifier = x.get("identifier", "N/A")

            f.write(f"Flowsheet {idx}\n")
            f.write("-" * 120 + "\n")

            f.write(f"identifier: {identifier}, ")
            f.write(f"situation index: {pi.get('feed_situation_index')}, ")
            f.write(f"components in feed: {pi.get('indices_components_in_feeds')}, ")
            f.write(f"feeds: {pi.get('list_feed_streams')}, ")
            f.write(f"npv_normed: {x.get('obj')}, ")
            f.write(f"per_ratio: {x.get('per_ratio')}, ")
            f.write(f"npv_wo_app_cost: {x.get('npv_wo_app_cost')}, ")
            f.write(f"npv_raw: {x.get('npv_raw')}, ")
            f.write(f"total_units_placed: {x.get('total_units_placed')}\n")

            f.write("\nNodes:\n")

            for node_idx, node_data in graph._node.items():
                f.write(f"  node {node_idx}:\n")

                for k, v in node_data.items():

                    if isinstance(v, np.ndarray):
                        v = v.tolist()

                    f.write(f"    {k}: {v}\n")

            f.write("\nEdges:\n")

            for src, dst, key, data in graph.edges(keys=True, data=True):

                f.write(f"\nedge {src} -> {dst}\n")

                f.write(
                    f"  output_label : {data.get('output_label')}, "
                )

                f.write(
                    f"  is_recycle   : {data.get('is_recycle')}, "
                )

                flow = data.get("stream", {}).get("flow")

                if flow is not None:

                    if isinstance(flow, np.ndarray):
                        flow = flow.tolist()

                    f.write(
                        f"  flow         : {flow}"
                    )

            f.write("\n\n")
            f.write("=" * 120 + "\n\n")


dump_pickle_flowsheets_txt(
    pickle_path="/home/akhanna/GumbelFlowsheets/complete_generated_flowsheets.pickle",
    txt_path="/home/akhanna/GumbelFlowsheets/generated_flowsheets_complete_butan_water.txt",
)
```

```python
import re
import numpy as np
import pandas as pd


def analyze_flowsheet_txt(filepath):
    """
    Analyze flowsheets stored in a txt logfile.

    Difficulty is defined relative to each subsystem's test composition:

        ethanol_water      -> [0.5, 0.5, 0.0]
        acetone_chloro     -> [0.5, 0.5, 0.0]
        water_py           -> [0.9, 0.1, 0.0]
        n-butanol_water    -> [0.6, 0.4, 0.0]

    Categories:
        test   : exactly the test composition
        hard   : distance from test <= 0.15
        medium : distance from test <= 0.40
        easy   : distance from test > 0.40

    Distance is measured using the first feed component, since the
    binary feed compositions sum to 1.

    Returns
    -------
    df : pd.DataFrame
        One row per flowsheet.

    subsystem_stats : pd.DataFrame
        Overall statistics for each subsystem.

    category_stats : pd.DataFrame
        Statistics for each subsystem and difficulty category.
    """

    # =========================================================
    # Test compositions
    # =========================================================

    test_compositions = {
        "ethanol_water": 0.50,
        "acetone_chloroform": 0.50,
        "water_pyridine": 0.90,
        "n-butanol_water": 0.60,
    }

    # =========================================================
    # Regex patterns
    # =========================================================

    identifier_pattern = re.compile(
        r"identifier:\s*([^,]+)"
    )

    subsystem_pattern = re.compile(
        r"subsystem:\s*([^,]+)"
    )

    obj_pattern = re.compile(
        r"obj:\s*([-+eE0-9.]+)"
    )

    npv_pattern = re.compile(
        r"npv_normed:\s*([-+eE0-9.]+)"
    )

    feed_pattern = re.compile(
        r"feeds:\s*\[array\(\[\s*"
        r"([-+eE0-9.]+)\s*,\s*"
        r"([-+eE0-9.]+)\s*,\s*"
        r"([-+eE0-9.]+)"
    )

    # =========================================================
    # Parse txt file
    # =========================================================

    records = []
    current_round = None

    with open(filepath, "r") as f:

        for line in f:

            line = line.strip()

            # -------------------------------------------------
            # Get round number
            # -------------------------------------------------

            if line.startswith("round:"):

                try:
                    current_round = int(
                        line.split(":")[1].strip()
                    )
                except ValueError:
                    current_round = None

                continue

            # -------------------------------------------------
            # Main flowsheet header
            # -------------------------------------------------

            if not line.startswith("identifier:"):
                continue

            identifier_match = identifier_pattern.search(line)
            subsystem_match = subsystem_pattern.search(line)
            obj_match = obj_pattern.search(line)
            npv_match = npv_pattern.search(line)
            feed_match = feed_pattern.search(line)

            if not all([
                identifier_match,
                subsystem_match,
                obj_match,
                npv_match,
                feed_match
            ]):
                print("\nCould not parse:")
                print(line)
                continue

            identifier = identifier_match.group(1).strip()
            subsystem = subsystem_match.group(1).strip()

            obj = float(
                obj_match.group(1)
            )

            npv_normed = float(
                npv_match.group(1)
            )

            feed_1 = float(
                feed_match.group(1)
            )

            feed_2 = float(
                feed_match.group(2)
            )

            feed_3 = float(
                feed_match.group(3)
            )

            records.append(
                {
                    "round": current_round,
                    "identifier": identifier,
                    "subsystem": subsystem,
                    "obj": obj,
                    "npv_normed": npv_normed,
                    "feed_1": feed_1,
                    "feed_2": feed_2,
                    "feed_3": feed_3,
                }
            )

    # =========================================================
    # Convert to dataframe
    # =========================================================

    df = pd.DataFrame(records)

    if len(df) == 0:
        raise ValueError(
            "No flowsheets could be parsed from the file."
        )

    # =========================================================
    # Distance from subsystem test composition
    # =========================================================

    def get_distance_from_test(row):

        subsystem = row["subsystem"]

        if subsystem not in test_compositions:
            return np.nan

        test_x = test_compositions[subsystem]

        return abs(
            row["feed_1"] - test_x
        )

    df["distance_from_test"] = df.apply(
        get_distance_from_test,
        axis=1
    )

    # =========================================================
    # Categorize difficulty
    # =========================================================

    def categorize(row):

        subsystem = row["subsystem"]

        if subsystem not in test_compositions:
            return "uncategorized"

        distance = row["distance_from_test"]

        # Exact test composition
        if np.isclose(
            distance,
            0.0,
            atol=1e-6
        ):
            return "test"

        # Within 0.15 of test case
        elif distance <= 0.15:
            return "hard"

        # Between 0.15 and 0.40
        elif distance <= 0.40:
            return "medium"

        # More than 0.40 away
        else:
            return "easy"

    df["category"] = df.apply(
        categorize,
        axis=1
    )

    # =========================================================
    # OVERALL SUBSYSTEM STATISTICS
    # =========================================================

    subsystem_stats = (
        df
        .groupby("subsystem")
        .agg(

            # Number of flowsheets
            num_instances=(
                "obj",
                "size"
            ),

            # Number where NPV > 0.95
            num_npv_above_090=(
                "npv_normed",
                lambda x: (x > 0.90).sum()
            ),

            # Objective statistics
            min_obj=(
                "obj",
                "min"
            ),

            max_obj=(
                "obj",
                "max"
            ),

            mean_obj=(
                "obj",
                "mean"
            ),

            median_obj=(
                "obj",
                "median"
            ),

            std_obj=(
                "obj",
                "std"
            ),
        )
        .reset_index()
    )

    # Percentage with NPV > 0.95
    subsystem_stats["percent_npv_above_090"] = (
        subsystem_stats["num_npv_above_090"]
        / subsystem_stats["num_instances"]
        * 100
    )

    # =========================================================
    # CATEGORY STATISTICS
    # =========================================================

    category_stats = (
        df
        .groupby(
            [
                "subsystem",
                "category"
            ]
        )
        .agg(

            num_instances=(
                "obj",
                "size"
            ),

            # Objective range
            min_obj=(
                "obj",
                "min"
            ),

            max_obj=(
                "obj",
                "max"
            ),

            # Additional useful statistics
            mean_obj=(
                "obj",
                "mean"
            ),

            median_obj=(
                "obj",
                "median"
            ),

            std_obj=(
                "obj",
                "std"
            ),

            # NPV > 0.95
            num_npv_above_090=(
                "npv_normed",
                lambda x: (x > 0.90).sum()
            ),
        )
        .reset_index()
    )

    category_stats["percent_npv_above_090"] = (
        category_stats["num_npv_above_090"]
        / category_stats["num_instances"]
        * 100
    )

    # =========================================================
    # Make category ordering nice
    # =========================================================

    category_order = [
        "test",
        "hard",
        "medium",
        "easy",
        "uncategorized"
    ]

    category_stats["category"] = pd.Categorical(
        category_stats["category"],
        categories=category_order,
        ordered=True
    )

    category_stats = category_stats.sort_values(
        [
            "subsystem",
            "category"
        ]
    ).reset_index(drop=True)

    # =========================================================
    # PRINT RESULTS
    # =========================================================

    print("\n")
    print("=" * 100)
    print("OVERALL SUBSYSTEM STATISTICS")
    print("=" * 100)

    print(
        subsystem_stats.to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}"
        )
    )

    print("\n")
    print("=" * 100)
    print("OBJECTIVE STATISTICS BY DIFFICULTY")
    print("=" * 100)

    print(
        category_stats.to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}"
        )
    )

    # =========================================================
    # Print feed ranges for sanity checking
    # =========================================================

    print("\n")
    print("=" * 100)
    print("CATEGORY FEED RANGES FOUND IN FILE")
    print("=" * 100)

    feed_ranges = (
        df
        .groupby(
            [
                "subsystem",
                "category"
            ],
            observed=True
        )
        .agg(
            min_feed_1=("feed_1", "min"),
            max_feed_1=("feed_1", "max"),
            num_instances=("feed_1", "size"),
        )
        .reset_index()
    )

    print(
        feed_ranges.to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}"
        )
    )

    return df, subsystem_stats, category_stats


df, subsystem_stats, category_stats = analyze_flowsheet_txt(
    "/home/akhanna/GumbelFlowsheets/environ_correct/all/top_train_flowsheets.txt"
)

'''print(
    df[
        (df["subsystem"] == "ethanol_water")
        & (df["category"] == "hard")
    ]
)'''
```

```python
import re
import numpy as np
import pandas as pd
from pathlib import Path


# Reference test instances for each situation
TEST_COMPOSITIONS = {
    0: {
        "name": "acetone_chloroform",
        "composition": np.array([0.5, 0.5, 0.0]),
    },
    1: {
        "name": "ethanol_water",
        "composition": np.array([0.5, 0.5, 0.0]),
    },
    2: {
        "name": "n_butanol_water",
        "composition": np.array([0.6, 0.4, 0.0]),
    },
    3: {
        "name": "water_py",
        "composition": np.array([0.9, 0.1, 0.0]),
    },
}


def categorize_distance(distance, is_test, hard_threshold=0.15,
                        medium_threshold=0.40):
    """
    Assign a difficulty category based on distance from the test composition.
    """
    if is_test:
        return "test"
    elif distance <= hard_threshold:
        return "hard"
    elif distance <= medium_threshold:
        return "medium"
    else:
        return "easy"


def calculate_stats(df):
    """
    Calculate requested objective/NPV statistics for a DataFrame.
    """
    if len(df) == 0:
        return {
            "num_instances": 0,
            "num_npv_above_090": 0,
            "min_obj": np.nan,
            "max_obj": np.nan,
            "mean_obj": np.nan,
            "median_obj": np.nan,
            "std_obj": np.nan,
            "percent_npv_above_090": np.nan,
        }

    obj = df["objective"]

    num_above = (obj > 0.90).sum()

    return {
        "num_instances": len(obj),
        "num_npv_above_090": int(num_above),
        "min_obj": obj.min(),
        "max_obj": obj.max(),
        "mean_obj": obj.mean(),
        "median_obj": obj.median(),

        # Population standard deviation.
        # Change ddof=1 if you want sample std.
        "std_obj": obj.std(ddof=0),

        "percent_npv_above_090": 100.0 * num_above / len(obj),
    }


def analyze_results_file(
    file_path,
    hard_threshold=0.15,
    medium_threshold=0.40,
    npv_threshold=0.90,
):
    """
    Parse a results .txt file and compute statistics by mixture
    and difficulty category.

    Parameters
    ----------
    file_path : str or Path
        Path to the results text file.

    hard_threshold : float
        Maximum Euclidean distance for the 'hard' category.

    medium_threshold : float
        Maximum Euclidean distance for the 'medium' category.

    npv_threshold : float
        Threshold used for counting high-NPV instances.

    Returns
    -------
    games_df : pandas.DataFrame
        One row per parsed game.

    overall_stats : pandas.DataFrame
        Statistics for each mixture using all instances.

    category_stats : pandas.DataFrame
        Statistics for each mixture broken down into
        test / hard / medium / easy.
    """

    text = Path(file_path).read_text()

    # ---------------------------------------------------------
    # Split file into individual games
    # ---------------------------------------------------------
    game_blocks = re.split(
        r"(?=game\s+nr\s*:\s*\d+)",
        text,
        flags=re.IGNORECASE
    )

    records = []

    for block in game_blocks:
        if not re.search(r"game\s+nr\s*:", block, re.IGNORECASE):
            continue

        # -----------------------------------------------------
        # Game number
        # -----------------------------------------------------
        game_match = re.search(
            r"game\s+nr\s*:\s*(\d+)",
            block,
            re.IGNORECASE
        )

        # -----------------------------------------------------
        # Winning objective
        # -----------------------------------------------------
        obj_match = re.search(
            r"winning\s+obj\s*>>>\s*"
            r"([-+]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][-+]?\d+)?)",
            block,
            re.IGNORECASE
        )

        # -----------------------------------------------------
        # Situation index
        # -----------------------------------------------------
        situation_match = re.search(
            r"situation\s+index\s*:\s*(\d+)",
            block,
            re.IGNORECASE
        )

        # -----------------------------------------------------
        # Feed composition
        #
        # Handles:
        # feeds: [array([0.5, 0.5, 0. ])]
        # -----------------------------------------------------
        feed_match = re.search(
            r"feeds\s*:\s*\[\s*array\s*\(\s*\[([^\]]+)\]",
            block,
            re.IGNORECASE
        )

        if not all([
            game_match,
            obj_match,
            situation_match,
            feed_match
        ]):
            print(
                "Warning: could not fully parse a game block. "
                "Skipping it."
            )
            continue

        game_nr = int(game_match.group(1))
        objective = float(obj_match.group(1))
        situation_index = int(situation_match.group(1))

        # Parse things such as
        # '0.5, 0.5, 0.'
        feed = np.fromstring(
            feed_match.group(1),
            sep=","
        )

        if situation_index not in TEST_COMPOSITIONS:
            print(
                f"Warning: unknown situation index "
                f"{situation_index} in game {game_nr}. Skipping."
            )
            continue

        mixture_info = TEST_COMPOSITIONS[situation_index]

        mixture_name = mixture_info["name"]
        test_composition = mixture_info["composition"]

        if len(feed) != len(test_composition):
            print(
                f"Warning: composition dimension mismatch "
                f"in game {game_nr}. Skipping."
            )
            continue

        # -----------------------------------------------------
        # Euclidean distance to the reference test composition
        # -----------------------------------------------------
        distance = np.linalg.norm(feed - test_composition)

        # Use isclose rather than == because these are floats.
        is_test = np.allclose(
            feed,
            test_composition,
            rtol=0,
            atol=1e-10
        )

        category = categorize_distance(
            distance,
            is_test=is_test,
            hard_threshold=hard_threshold,
            medium_threshold=medium_threshold,
        )

        records.append({
            "game_nr": game_nr,
            "situation_index": situation_index,
            "mixture": mixture_name,

            "x1": feed[0],
            "x2": feed[1],
            "x3": feed[2],

            "test_x1": test_composition[0],
            "test_x2": test_composition[1],
            "test_x3": test_composition[2],

            "distance_from_test": distance,
            "category": category,
            "objective": objective,
        })

    games_df = pd.DataFrame(records)

    if games_df.empty:
        raise ValueError(
            "No games could be parsed from the file. "
            "Check the file format."
        )

    # ---------------------------------------------------------
    # Internal statistics helper using requested NPV threshold
    # ---------------------------------------------------------
    def stats(group):
        obj = group["objective"]

        num_instances = len(obj)
        num_above = int((obj > npv_threshold).sum())

        return pd.Series({
            "num_instances": num_instances,
            "num_npv_above_090": num_above,
            "min_obj": obj.min(),
            "max_obj": obj.max(),
            "mean_obj": obj.mean(),
            "median_obj": obj.median(),
            "std_obj": obj.std(ddof=0),
            "percent_npv_above_090":
                100.0 * num_above / num_instances,
        })

    # ---------------------------------------------------------
    # Overall statistics for each mixture
    # ---------------------------------------------------------
    overall_stats = (
        games_df
        .groupby(
            ["situation_index", "mixture"],
            observed=True
        )
        .apply(stats, include_groups=False)
        .reset_index()
    )

    # ---------------------------------------------------------
    # Statistics for test / hard / medium / easy
    # ---------------------------------------------------------
    category_order = ["test", "hard", "medium", "easy"]

    games_df["category"] = pd.Categorical(
        games_df["category"],
        categories=category_order,
        ordered=True
    )

    category_stats = (
        games_df
        .groupby(
            ["situation_index", "mixture", "category"],
            observed=True
        )
        .apply(stats, include_groups=False)
        .reset_index()
        .sort_values(
            ["situation_index", "category"]
        )
        .reset_index(drop=True)
    )

    return games_df, overall_stats, category_stats


games, overall_stats, category_stats = analyze_results_file(
    "/home/akhanna/GumbelFlowsheets/evaluation_prints_alternative.txt"
)

print("\n=== OVERALL STATS ===")
print(overall_stats.to_string(index=False))

print("\n=== STATS BY DIFFICULTY ===")
print(category_stats.to_string(index=False))
```
